## The business model: an OTA built around AirAsia flights

AirAsia MOVE is an online travel agency (OTA) and lifestyle app: flight bookings (AirAsia and increasingly third-party carriers, including new partnerships with VietJet and IndiGo), hotel stays, ride-hailing, and duty-free shopping. Its core commercial dependency, flagged already in Notebook 4's ownership map, is that even as a legally separate Capital A business, its transaction volume is structurally tied to how many seats AirAsia Group actually sells. **[PUBLIC, Capital A financial results press releases]**

## The honest trend: revenue and EBITDA both declined

Revenue fell from **RM689 million (FY2023)** to **RM641 million (FY2025)**, a real 7% decline, and EBITDA fell more sharply, from **RM137 million** to **RM84 million**, a 39% drop. This is the one business among Capital A's five where the multi-year trend is genuinely negative, not a case of decelerating growth like Teleport or ADE. **[PUBLIC, Capital A FY2023 and FY2025 financial results]**

The likely driver, based on this project's own earlier finding (Notebook 4): AirAsia MOVE was previously reported as "soft," attributed directly to "airline capacity moderation," the company's own words, and the airline's own 2Q26 capacity was down 11% year-on-year (Notebook 5) as part of its fuel-shock response. A travel app whose core product is booking AirAsia flights has less to sell when AirAsia itself is flying fewer seats. Full-year FY2024 figures for MOVE specifically were not disclosed cleanly enough at this level to include here, a genuine data gap rather than an omission.

There is a real turnaround signal within the FY2025 numbers, though: 4Q2025 revenue reached **RM300 million**, nearly triple the prior quarter and nearly double the same quarter a year earlier, driven by the launch of a new B2B booking business that alone contributed roughly 55% of that quarter's revenue. **[PUBLIC, AirAsia Group 2Q26/4Q25 disclosures]** That means most of FY2025's weakness was concentrated in the first three quarters, with a genuinely strong finish, a pattern the full-year total alone understates.

## The decline reversed in 2026, and a genuine reporting quirk worth disclosing

The two-year decline did not continue into 2026. For **2Q2026**, AirAsia MOVE grew revenue **5% year-on-year to RM96 million** and remained **EBITDA-positive at RM7 million**, the company's own language describing this as "operational resilience... despite travel industry headwinds." For **1Q2026**, the company had separately reported EBITDA of RM14.1 million (a 12% year-on-year dip at the time). On a first-half (1H2026) basis, the company states revenue of **RM194 million**, EBITDA of **RM14 million**, and NOP of **RM4.9 million**. **[PUBLIC, Capital A First Quarter 2026 and Second Quarter 2026 financial results press releases]**

Worth flagging plainly: the disclosed 1Q26 EBITDA (RM14.1 million) plus the disclosed 2Q26 EBITDA (RM7 million) sum to roughly RM21 million, not the RM14 million Capital A states as the 1H26 total. This project cannot fully reconcile that gap from the two press releases alone. Capital A's own 1Q26 release notes that some of its segmental figures are shown "on a pre-elimination basis," which may explain part of the difference (inter-segment transactions being netted out at the half-year consolidated level), but this project has not confirmed that this specific gap is fully explained by that note. Rather than pick whichever number tells a cleaner story, both figures are reported here as disclosed. **[Disclosed, unreconciled discrepancy]**

## Explain It Simply

AirAsia MOVE is the travel booking app, and it's the one part of Capital A's business that actually got smaller over its first two full years under Capital A: both revenue and profit fell. The most likely reason is straightforward: the app mostly sells AirAsia flights, and AirAsia itself was flying fewer seats as part of its response to expensive fuel, so there was simply less for the app to sell. The good news is that the decline reversed in 2026, with growth returning in both quarters reported so far. Along the way, this project also found a real, unresolved wrinkle in the company's own numbers, its two quarterly profit figures for early 2026 don't add up to the half-year total the company separately reported, and rather than guess which number is right, both are shown here honestly. Unlike the maintenance and logistics businesses, there's no fair way to compare AirAsia MOVE against competitors like Traveloka or Klook, since neither publishes financial results.

## Glossary

- **OTA (Online Travel Agency)**: a platform that sells travel bookings (flights, hotels, and related services) on behalf of multiple providers.
- **B2B (Business-to-Business)**: here, a booking product sold to corporate or wholesale travel buyers rather than individual consumers.

## Why Is This Happening?

**Why would a legally separate company's revenue move in lockstep with its former parent's capacity decisions?** Legal separation, completed in the January 2026 restructuring (Notebook 3), changes who bears financial and legal liability, but it does not sever a genuine commercial dependency. AirAsia MOVE's core inventory (flights to sell) is not something it controls; it is downstream of AirAsia Group's own capacity choices, which are themselves downstream of the fuel-shock response (Notebook 5). This is the clearest example in this project of the coupling risk flagged in Notebook 4: two separate companies can still transmit real business pressure to each other through ordinary commercial dependency, without any change in ownership or contagion in the stock market sense.
