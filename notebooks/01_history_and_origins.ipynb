# 01: History and Origins

**Business question**: Before diagnosing whether AirAsia's September 2026 crisis is a genuine solvency risk or an overreaction, it helps to understand how AirAsia and its long-haul affiliate AirAsia X got here. Both the RM1 buyout that built the group and the near-death experiences it survived along the way shape how seriously to take today's numbers.

## From a government-linked carrier to a RM1 buyout

AirAsia was founded in **1993** as a government-linked domestic carrier, years before Tony Fernandes was ever involved. **[PUBLIC]**

In **December 2001**, Tony Fernandes and Kamarudin Meranun, through their company Tune Air Sdn Bhd, acquired the loss-making airline for a nominal **RM1**, taking on approximately **RM40 million** in existing debt. This is the founding story the company itself tells, and it is corroborated across independent sources. **[PUBLIC]**

The tagline "Now Everyone Can Fly" originates from this 2001 relaunch, not from AirAsia X specifically, reflecting Fernandes's stated vision of making air travel affordable and accessible in a region where flying had been a premium product. **[PUBLIC]**

## The long-haul affiliate: AirAsia X

AirAsia X was established in **2006** as FlyAsianXpress Sdn Bhd, rebranded AirAsia X in **2007**, and flew its first international route from Kuala Lumpur to Gold Coast, Australia in **November 2007**. It was set up as a separate legal entity from AirAsia Berhad specifically to run long-haul, low-cost routes, a structural choice that matters later: it is this same AirAsia X entity that ends up as the legal acquirer in the January 2026 reverse acquisition (Notebook 3). **[PUBLIC]**

## Who is actually the mother company here

This is worth being precise about, since the group's later restructuring inverts it. **AirAsia Group Berhad, the 1993 company Tony Fernandes bought for RM1, is the original parent.** It later renamed itself **Capital A Berhad in January 2022** (Notebook 2), but it is the same continuous legal entity: today's Capital A is that original mother company, just under a different name. **AirAsia X Berhad was never a subsidiary of it**; it was founded independently in 2006/2007 as its own separately listed sister company. **[PUBLIC]**

That ownership picture flips in 2026. AirAsia X legally acquires the group's short/medium-haul airlines from Capital A (Notebook 3), becomes the parent of the combined aviation business, and then renames *itself* "AirAsia Group Berhad" in July 2026, reusing the original parent's old name for a legally different company. From that point on, "Capital A" (the true original mother) and "AirAsia Group" (now legally the former sister company) are two separate listed businesses linked by a minority cross-shareholding, not one still sitting inside the other (Notebook 4 maps this fully).

## Explain It Simply

AirAsia was not always Tony Fernandes's company. It started as a small, loss-making, government-linked airline in 1993. In 2001, Fernandes and his business partner bought it for just RM1, took on its debts, and turned it into the low-cost airline that made "Now Everyone Can Fly" a household phrase. That original company later renamed itself Capital A. A separate long-haul sister airline, AirAsia X, was created in 2006 to fly longer international routes under the same brand but as its own legal company, never owned by the original parent. For most of their history these two airlines ran side by side, not as one company. Then in 2026, the relationship flipped: AirAsia X bought the original company's airlines and took over the "AirAsia Group" name for itself, while the original company (now Capital A) kept everything else and became a minority shareholder in the very airline group it used to own outright.

## Glossary

- **Tune Air Sdn Bhd**: the private investment vehicle Tony Fernandes and Kamarudin Meranun used to acquire AirAsia in 2001.
- **AirAsia X**: the long-haul, low-cost affiliate founded in 2006/2007, legally separate from AirAsia Berhad for most of its history.
- **PN17**: Bursa Malaysia's Practice Note 17, a classification for financially distressed listed companies, covered in depth in Notebook 2.

## Why Is This Happening?

**Why did AirAsia and AirAsia X stay separate legal entities for two decades instead of merging earlier?** Splitting short/medium-haul (AirAsia) from long-haul (AirAsia X) let each business raise its own financing and bear its own risk, long-haul routes have very different aircraft, fuel, and crew economics from short-haul ones, so keeping them in separate listed entities let investors judge each on its own merits rather than one business's troubles dragging down the other's share price. That logic held for two decades. The 2026 merger reversed it, betting that a single combined entity is now more efficient than two separately financed ones, a bet this project tests directly in Notebook 3.
