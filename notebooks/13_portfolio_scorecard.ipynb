## An honest caveat: margin comparisons across very different business models

EBITDA margin is not a perfectly fair comparison across six structurally different businesses. AirAsia Next's 33.3% margin partly reflects a licensing-heavy, asset-light model with few direct costs, not necessarily better execution than ADE's 22% margin, which is earned running physical hangars and maintenance crews. Airlines structurally run thinner margins than most of these business types even when healthy, so the airline's 9% margin sitting mid-pack does not mean its situation is comparable to Santan's 4.4%, the airline's problem is the negative revenue growth alongside real balance-sheet stress (Notebook 14), not the margin figure alone. **[Disclosed limitation]**

## Explain It Simply

Putting every part of this business empire side by side on the exact same three months tells a clear story: the airline is the problem child, not the group as a whole. Every other business, the maintenance arm, the logistics business, the fintech and loyalty brand, the travel app, and even the catering business, grew its revenue in the same quarter the airline's shrank. The airline isn't necessarily the least profitable as a percentage of revenue (a couple of the smaller businesses run thinner margins), but it's the only one actually going backwards on the top line, and it's the one with the real balance-sheet stress documented elsewhere in this project. If you only read headlines about "AirAsia's crisis," you'd assume the whole empire was struggling. The reality, backed by numbers from the exact same quarter, is that one piece, the biggest piece, is in real trouble, while the businesses built around it are growing.

## Glossary

- **Scorecard**: a side-by-side comparison of multiple business units on the same set of metrics, used to spot which are outperforming or underperforming the group as a whole.
- **EBITDA margin**: EBITDA expressed as a percentage of revenue, a measure of how much of each ringgit of revenue converts to operating profit before interest, tax, depreciation, and amortisation.
- **Normalised**: adjusted to remove a specific distorting factor (here, Santan's flat 2Q26 revenue is stated on a basis that removes some one-off effect the company did not detail further).

## Why Is This Happening?

**Why would the airline deteriorate so much more sharply than the businesses built around it?** The airline is directly exposed to the fuel price shock (Notebook 5) in a way none of Capital A's five retained businesses are: fuel is a direct, large line item in the airline's own cost structure, while Teleport, ADE, AirAsia MOVE, AirAsia Next, and Santan are exposed to the airline's troubles only indirectly, through reduced flight capacity affecting their own volumes (Notebooks 10 and 12), a real but much smaller and more diffuse effect. The 2026 split (Notebook 3) therefore did what it was structurally supposed to do: it concentrated aviation-specific risk (fuel prices, aircraft economics, PN17 history) inside the airline entity, while giving the retained businesses a real, if imperfect, degree of insulation from that specific risk. This project's own evidence supports treating the crisis as company-specific to the airline, not evidence that Capital A's broader strategic bet on non-aviation businesses has failed.
