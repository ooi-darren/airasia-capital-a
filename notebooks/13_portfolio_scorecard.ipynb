## The headline finding: the airline, not the rest of the group, is where the real deterioration is

Comparing each unit's most recent year-on-year quarter (1Q2026 for Capital A's five businesses, 2Q2026 for the airline, the most recent period each disclosed): **ADE grew revenue 7.4% and EBITDA 7%**, both improving. **Teleport grew revenue 20%**, with EBITDA down 6% in ringgit terms but up 6% on a constant-currency basis, a currency effect rather than a genuine deterioration (Notebook 8). **AirAsia MOVE's revenue fell 6% and EBITDA fell 12%**, a real but modest softening, still profitable at the PAT level (Notebook 10). **AirAsia Next (-5% revenue) and Santan (-3.8% revenue)** also softened, without a company-quantified profit change to plot here (Notebooks 11-12).

**The airline, by contrast, saw revenue roughly flat (-1%) but EBITDA collapse 56%, swinging to a net operating loss (-230% YoY)**, by far the most severe deterioration of any business examined in this entire project. **[PUBLIC, Capital A 1Q2026 and AirAsia Group 2Q26 disclosures, both extracted directly by this project]**

This directly answers the business question the original restructuring of this project was built to test: the September 2026 crisis is not a symptom of a broadly struggling ecosystem. It is concentrated almost entirely in one business, the airline, while the five businesses Capital A retained after the 2026 split are mostly stable to modestly softening, with two (ADE, Teleport) genuinely improving on both growth and profitability.

## An honest caveat: different quarters, not a perfectly matched snapshot

This scorecard compares Capital A's five businesses on 1Q2026 (their most recent disclosed quarter at the time of this research) against the airline's 2Q2026 (its most recent disclosed quarter). ## Explain It Simply

Putting every part of this business empire side by side on their most recent numbers tells a clear story: the airline is the problem child, not the group as a whole. Two of Capital A's five retained businesses, the maintenance arm (ADE) and the logistics business (Teleport), are genuinely growing and getting more profitable. Three others, the travel app, the brand/fintech business, and the catering arm, are softening a bit, mostly because they depend somewhat on the airline flying more planes. But none of that softening comes close to what happened at the airline itself, where profit collapsed by more than half and flipped into an actual operating loss, while revenue barely moved. If you only read headlines about "AirAsia's crisis," you'd assume the whole empire was struggling. The reality, backed by the actual numbers, is that one piece, the biggest piece, is in real trouble, while the businesses built around it are mostly fine or even doing well.

## Glossary

- **Scorecard**: a side-by-side comparison of multiple business units on the same set of metrics, used to spot which are outperforming or underperforming the group as a whole.
- **Constant-currency basis**: a growth figure recalculated to remove the effect of exchange rate movements, isolating the underlying business performance from currency swings.

## Why Is This Happening?

**Why would the airline deteriorate so much more sharply than the businesses built around it?** The airline is directly exposed to the fuel price shock (Notebook 5) in a way none of Capital A's five retained businesses are: fuel is a direct, large line item in the airline's own cost structure, while Teleport, ADE, AirAsia MOVE, AirAsia Next, and Santan are exposed to the airline's troubles only indirectly, through reduced flight capacity affecting their own volumes (Notebooks 10 and 12), a real but much smaller and more diffuse effect. The 2026 split (Notebook 3) therefore did what it was structurally supposed to do: it concentrated aviation-specific risk (fuel prices, aircraft economics, PN17 history) inside the airline entity, while giving the retained businesses a real, if imperfect, degree of insulation from that specific risk. This project's own evidence supports treating the crisis as company-specific to the airline, not evidence that Capital A's broader strategic bet on non-aviation businesses has failed.
