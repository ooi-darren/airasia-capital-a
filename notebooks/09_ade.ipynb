**Business question**: ADE (Asia Digital Engineering) is Capital A's aircraft maintenance, repair and overhaul (MRO) business, and unlike most of Capital A's other units, it competes in a market with genuinely comparable, publicly-listed competitors. Is it actually competitive at that scale?

## The business model: captive demand, expanding to third parties

ADE performs base and line maintenance, primarily for AirAsia Group's own fleet, while actively expanding into third-party work for other airlines (a stated contract with Air France, secured in 2025). It has been growing hangar capacity, including a new hangar in Senai and one in Sepang with 14 additional lines, plus a landing gear joint venture with GMF. **[PUBLIC, Capital A financial results press releases]**

## Three years of real trend: consistent growth, a genuine EBITDA dip and recovery

Revenue grew from **RM574 million (FY2023)** to **RM727 million (FY2024, +27% YoY)** to **RM895 million (FY2025)**, a real 56% increase over two years, though the growth rate decelerated as the base grew, a normal pattern for a maturing business rather than a warning sign on its own. EBITDA held roughly flat between FY2023 (**RM146 million**) and FY2024 (approximately RM138 million, derived from the disclosed 19% margin, since the FY2024 press release stated only the margin directly), before jumping to **RM205 million in FY2025**. **[PUBLIC, Capital A FY2023-FY2025 financial results; FY2024 EBITDA absolute figure DERIVED from disclosed margin]**

FY2024's EBITDA margin compression (down 6 percentage points to 19%) was attributed to higher staff costs and upfront costs from the new 14-line Sepang hangar, a genuine one-off investment cost rather than a sign of a weakening core business, consistent with FY2025's strong recovery once that capacity came online.

## The strongest revenue growth of any Capital A business in 2Q2026

ADE's momentum accelerated further into 2026. For **2Q2026**, revenue reached **RM284 million, a 29% year-on-year increase**, the fastest revenue growth rate of any of Capital A's five businesses that quarter, with EBITDA of **RM69 million at a 22% margin**, the highest margin among ADE, Teleport, AirAsia MOVE, and Santan that quarter (Notebook 13's portfolio scorecard shows AirAsia Next's licensing-heavy model runs a higher margin still). **SIA Engineering Company**, Singapore Airlines' publicly-listed MRO subsidiary, is a genuinely comparable competitor, since both are Southeast Asian aircraft MRO providers serving both a captive parent airline and third-party customers. For its FY2025-26 (financial year ended around March 2026), SIA Engineering reported **S$1.245 billion revenue (+13.8% YoY)**. **[PUBLIC, SIA Engineering Group full-year results]**

## Explain It Simply

ADE fixes and maintains AirAsia's planes, and increasingly, other airlines' planes too. Its revenue grew a real 56% over two years. Along the way, in 2024, its profit margin briefly dipped because it was spending money building new maintenance hangar capacity, not because the underlying business got worse, and that investment paid off with a strong profit jump in 2025 once the new capacity came online. Compared to SIA Engineering, Singapore Airlines' much bigger maintenance arm, ADE is still about a quarter to a fifth of the size, a real gap, but nowhere near as extreme as some of Capital A's other businesses face against their own competitors.

## Glossary

- **MRO (Maintenance, Repair, and Overhaul)**: the aviation industry term for aircraft maintenance services.
- **Base maintenance**: major, scheduled aircraft maintenance work performed in a hangar, as opposed to routine checks between flights.
- **Line maintenance**: routine aircraft checks and minor repairs performed between flights, typically at the gate.

## Why Is This Happening?

**Why would a maintenance business deliberately compress its own margins to build new capacity?** Hangar capacity is a hard physical constraint on an MRO business's ability to take on more work, whether from its own captive fleet or third-party airlines. Investing in new hangar lines ahead of demand, accepting a temporary margin hit from staffing and setup costs, is a standard capital-intensive-industry trade-off: it sacrifices near-term profitability for the capacity needed to capture future growth, exactly the pattern the FY2024-to-FY2025 numbers show playing out.
