# 05: Where Capital A's Other Businesses Fit

**Business question**: After giving up its aviation businesses in January 2026, Capital A kept BigPay (fintech), Teleport (logistics), and AirAsia MOVE/Next/Santan (digital travel and lifestyle). Capital A's own shares fell 18% alongside AirAsia's 21% crash in September 2026, so did the split actually insulate the retained businesses, or is Capital A just as exposed?

## The retained business is genuinely profitable

For the quarter ended 30 June 2026, Capital A generated **RM809 million in revenue (+9% YoY)**, EBITDA of **RM108 million**, net operating profit of **RM38.5 million**, and profit after tax of **RM25 million**, its **second consecutive profitable quarter** since the aviation disposal. On a cumulative first-half basis, revenue reached **RM1.58 billion (+4% YoY)**. **[PUBLIC, Capital A Q2 2026 financial results]**

This is a real, positive signal: Capital A's retained non-aviation business is not simply a discard pile of leftover units, it is standing on its own feet financially, at least by the headline profit and revenue numbers.

**Teleport** (logistics/cargo) and **ADE** (Airasia Digital Enterprise) together generated over 70% of 1H26 group revenue, growing 21% and 19% YoY respectively. Teleport alone was Capital A's single largest revenue contributor, about 40% of group revenue in Q1 2026. **[PUBLIC]**

This is the real coupling risk: even after the legal and financial separation, Capital A's travel-linked digital businesses still depend on AirAsia Group actually flying planes. If AirAsia's capacity cuts (Notebook 7) continue or deepen, this is a plausible transmission channel from the aviation crisis into Capital A's own results, separate from any share-price panic contagion.

## A separate legal risk: the BigPay and Teleport share seizure

On **10 June 2026**, the Singapore High Court ordered the seizure of Move Digital's (a Capital A subsidiary) **99.56% stake in BigPay** and **11.45% stake in Teleport**, enforcing a December 2024 partial arbitration award. The dispute traces back to 2021, when BigPay's minority co-founders, Christopher Davison and Navin Rajagopalan, sought a buyout of their stakes under a 2017 shareholder agreement. They had sought between **US$140-183 million**; the arbitration tribunal awarded them only **US$14.736 million (RM59.85 million)**. Capital A/Move Digital stated it expects no losses from the seizure and planned to formally object to the enforcement action. **[PUBLIC]**

This is a genuine, disclosed legal risk sitting over exactly the two businesses (BigPay, Teleport) that matter most to Capital A's retained portfolio, Teleport being its largest revenue contributor. It predates the September 2026 aviation crisis by three months and is a separate, independent risk factor, not a symptom of it.

## Explain It Simply

When Capital A gave up its airlines in January 2026, it kept a logistics company (Teleport), a digital enterprise arm (ADE), a fintech app (BigPay), and some travel-related digital products (AirAsia MOVE, Next, Santan). The good news: this remaining business is actually profitable and growing, mostly thanks to Teleport and ADE. The less good news: the travel-related digital products are struggling specifically because AirAsia itself is flying fewer planes, so Capital A isn't fully insulated from AirAsia's troubles even though they're now separate companies. And separately, a Singapore court ordered Capital A's stake in BigPay and Teleport, its most valuable retained business, seized to enforce a legal payout to two BigPay co-founders, a real legal headache that has nothing to do with the aviation crisis but lands on the same company at almost the same time.

## Glossary

- **ADE**: Airasia Digital Enterprise, Capital A's digital enterprise/advertising and data business segment.
- **Teleport**: Capital A's air and ground logistics/cargo arm.
- **Arbitration award**: a binding decision from a private arbitration tribunal, enforceable through the courts like a judgment.

## Why Is This Happening?

**Why would AirAsia MOVE still be exposed to AirAsia Group's capacity cuts if the companies are now legally separate?** AirAsia MOVE's core product is a travel booking and lifestyle platform built around AirAsia flights; even as a separate legal entity, its transaction volume is structurally tied to how many seats AirAsia actually sells. Legal separation changes who bears financial and legal liability, but it does not sever a genuine commercial dependency, exactly the kind of distinction a headline like "Capital A and AirAsia are now separate companies" can obscure if taken to mean the businesses are fully independent of each other operationally.
