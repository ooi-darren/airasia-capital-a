# 11: Future, Scenarios and What's Next

**Business question**: Given everything established in this project, what does a realistic range of outcomes look like from here? This notebook is deliberately built as scenarios, not a forecast: this is a live, unresolved story, and no source consulted for this project predicts a specific outcome with confidence.

## The two levers that matter most

This project's own findings point to two variables that plausibly determine which direction this goes, more than any single announcement:

1. **Whether the refinancing succeeds.** AirAsia is seeking up to **USD1 billion** in international debt markets plus **RM700 million** in local credit facilities, explicitly for debt refinancing and balance-sheet consolidation, alongside a separate attempt to amend terms on an existing **USD200 million** private credit loan. **[PUBLIC]** Notebook 6's RM14.5 billion working-capital deficit is exactly the gap this financing is meant to bridge.
2. **Whether the fuel price shock persists or eases.** Notebook 2 showed the operating loss is fuel-driven, not a broad cost failure. A retreat in fuel prices would substantially improve the picture without any restructuring action at all; continued high prices would keep pressure on regardless of financing outcomes.

- **Refinancing succeeds, fuel eases**: the deficit gets bridged and the cost pressure that caused it recedes at the same time. The lightest-touch outcome, and consistent with the government's current lighter-touch posture (Notebook 10) proving to have been the right call.
- **Refinancing succeeds, fuel stays high**: the immediate liquidity crunch is addressed, but AirAsia keeps operating on thin or negative margins per seat-kilometre (Notebook 2) until fleet renewal (Notebook 7) or fare/cost actions catch up.
- **Refinancing stalls, partial route transfer**: the scenario the September 2026 reports describe, domestic routes only, contingent on the receiving carrier also taking the aircraft leases, not a full takeover.
- **Refinancing fails, full route absorption**: the most severe scenario, and the one furthest from anything currently reported as planned; no source used in this project describes this as an active plan, only as the kind of contingency being scenario-planned for.

## What this project's own evidence weighs toward

Without predicting an outcome, several of this project's own findings lean toward the less severe end of this range being more likely than the panic headlines implied: AirAsia Group has recent, direct precedent for surviving a far more severe restructuring (Notebook 3); the current operating loss is traceable to an external, potentially temporary fuel shock rather than structural mismanagement (Notebook 2); the government's revealed approach so far is lighter-touch than its own 2014 MAS precedent (Notebook 10); and Alton Aviation Consultancy's assessment, whose findings are not yet public, is explicitly a fact-finding step, not a decision. None of this rules out a harder outcome; it is the honest balance of evidence gathered here.

## Explain It Simply

Nobody, including this project, knows for certain what happens next to AirAsia. What can be said is that two things mostly decide it: whether AirAsia can raise the money it's trying to raise, and whether jet fuel prices come back down. If both go AirAsia's way, this crisis probably fades quietly. If neither does, the more severe scenarios discussed in the news, like other airlines taking over its routes, become more realistic. Based on everything gathered in this project, including the fact that this company has survived a much worse crisis before and that the current losses are traceable to an oil price shock rather than mismanagement, the balance of evidence leans toward this being serious but survivable, not necessarily toward the most alarming scenario.

## Glossary

- **Scenario analysis**: laying out a structured range of plausible futures based on identified variables, explicitly not the same as a forecast or probability estimate.

## Why Is This Happening?

**Why build scenarios instead of just picking the most likely one?** A single predicted outcome would misrepresent the actual state of the evidence: this is a live situation where the two decisive variables, refinancing success and fuel prices, are both genuinely uncertain and outside any single party's full control. Presenting a range tied explicitly to this project's own findings, rather than a single confident guess, is the more honest and more useful output for a reader trying to assess exposure or make a decision, exactly the standard this portfolio has held to since Notebook 1.
