## Correcting an earlier assumption

Direct extraction from AirAsia Group's own 2Q26 investor presentation shows the real figure is far smaller than the 30% initially assumed: the fleet went from 245 aircraft (2Q25) to 240 (1Q26) to 239 (2Q26), a reduction of only **6 aircraft, about 2.4%**. **[PUBLIC]**

## What is actually happening to the fleet

The company's own "Fleet Rightsizing" plan for 2026 is to return a total of **25 older aircraft** across the year, including **17 early returns** (ahead of their scheduled lease end dates), while receiving **2 new A321LR aircraft** (delivered April and June 2026) and expecting **3 more in 2027**. Separately, as of 2Q26: 1 aircraft was reactivated, 5 aircraft were undergoing checks (targeted to return to service 3Q-4Q26), and 1 widebody aircraft remained in storage for engine replacement. **[PUBLIC, AirAsia Group 2Q26 investor presentation]**

So the real story is a **gradual, planned fleet renewal and rightsizing programme**, returning older, higher-maintenance aircraft on favourable lease terms while phasing in newer, more fuel-efficient aircraft, not an emergency grounding of a third of the fleet. The 11% year-on-year ASK (capacity) cut identified in Notebook 2 comes more from reduced utilisation and route/frequency discipline ("held every route to strict profitability hurdle-rates," in the company's own words) than from a dramatically smaller fleet.

## A discrepancy inside the company's own disclosures, disclosed rather than hidden

Checking this more closely surfaced something worth flagging plainly: AirAsia Group's own 2Q26 **quarterly report narrative** (Note 27, "Review of Group Performance") states "approximately 30% of the Group's fleet (161 operating aircraft as at the end of the quarter) was non-operational." That figure does not appear anywhere else in the same company's own disclosures for the same quarter. The quarterly report's own operating statistics table lists 239 total aircraft with no operating/non-operating split. The separate 2Q26 **investor presentation** gives a specific breakdown instead: an "Active Fleet" of **231 of 239 aircraft (96.7%)**, with the gap accounted for by the 1 reactivated, 5 under checks, and 1 in storage detailed above, roughly **8 aircraft, about 3.3%**, not 30%. **[PUBLIC, both documents dated the same reporting period]**

This project relies on the investor presentation's figure, since it is the one backed by an itemised breakdown that adds up, rather than an unreconciled round number in the quarterly report's prose. But the honest position is that AirAsia's own disclosures contain two different answers to "what share of the fleet isn't flying," a full order of magnitude apart, for the same quarter, and this project cannot independently resolve which one the company actually meant. Readers checking the primary sources themselves should expect to find both numbers and should not assume this project picked the more flattering one without saying so.

## The capacity cuts are mostly about routes, not just aircraft

The 11% year-on-year ASK cut is not evenly spread. AirAsia Indonesia (IAA) and AirAsia Philippines (PAA), the two units the company itself flags as loss-making, had **33 unviable routes permanently cut**, bringing total permanent route closures to **64 for the Group year-to-date June 2026**. A further **17 routes were temporarily paused** across 2Q/3Q, with scheduling flexibility to restore them in late 2026 if conditions improve. Separately, the planned **Bahrain hub and a new London route were delayed to next year** to preserve capital. **[PUBLIC, AirAsia Group 2Q26 investor presentation]**

Guidance for the rest of the year cuts both ways: management is guiding capacity down a further **20-25% year-on-year for 3Q26** (the seasonally leanest quarter), before an explicit plan to **ramp capacity back up in 4Q26** toward pre-shock levels to capture year-end holiday demand. This is not a company retreating uniformly; it is concentrating cuts on specific underperforming markets and the leanest quarter, while planning to re-expand into the strongest one.

## Is this smart cost-cutting or shrinking the way out?

Returning older aircraft on favourable lease terms while bringing in more efficient replacements is, on its face, a reasonable response to a fuel price shock (Notebook 2): the aircraft being removed are disproportionately the ones with the worst fuel economy. The risk is that disciplined capacity cuts, if they persist too long or go too far, reduce the revenue base available to earn out of the working capital deficit identified in Notebook 6. Nothing in the disclosed plan suggests the cuts are currently at that scale; this is a real risk to monitor going forward, not a finding this notebook can confirm has already happened.

## Explain It Simply

An earlier piece of research for this project suggested AirAsia had grounded about a third of its planes, which would be a genuinely alarming sign of a company in freefall. Checking the actual company disclosures directly shows that claim mostly doesn't hold up: the fleet shrank by only 6 aircraft, about 2%, and the company's own investor presentation says only about 8 aircraft (3%) weren't flying at quarter-end. Oddly, the same company's own quarterly report text separately claims 30% wasn't flying, a number that doesn't match anything else the company disclosed for the same period; we're flagging that contradiction rather than quietly picking whichever number sounds better. What is really happening, and better documented, is a planned programme to retire older, less fuel-efficient planes early (25 of them across 2026) while bringing in newer ones, combined with cutting 64 underperforming routes (mostly in Indonesia and the Philippines) and guiding capacity down another 20-25% for the next quarter before ramping back up for year-end holidays. That's disciplined route and fleet management responding to expensive fuel, not a sign of a company shutting itself down.

## Glossary

- **Fleet rightsizing**: matching the number and type of aircraft to actual demand and route economics, often by returning leased aircraft early.
- **ASK (Available Seat Kilometres)**: total flying capacity, driven by both fleet size and how intensively that fleet is utilised.
- **IAA / PAA**: AirAsia Indonesia and AirAsia Philippines, the two country units the company itself identifies as loss-making and where most permanent route closures were concentrated.

## Why Is This Happening?

**Why return older aircraft specifically, rather than cutting capacity evenly across the fleet?** Older aircraft carry worse fuel economy and higher maintenance costs, precisely the two expense lines under the most pressure from the 2026 fuel shock (Notebook 2). Returning them early and replacing with newer aircraft (A321LR, and A220/A321XLR expected from 2028) is a way to cut the cost side of the unit-economics equation structurally, rather than simply flying fewer of the same planes. This is a genuinely different strategy from an across-the-board capacity freeze, and a more defensible one if the fuel shock persists.
