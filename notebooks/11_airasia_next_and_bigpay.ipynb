## The business model: brand licensing evolving into a real operating company

AirAsia Next started as a brand-licensing vehicle, monetising the AirAsia name (the "AirAsia Blue" licensing programme extends the brand beyond Southeast Asia, including into hotels). During FY2025 it became, in the company's own words, "a fully operational company with dedicated teams," and now also houses **BigPay**, the fintech and digital-wallet app, folding fintech and loyalty into what had been a pure licensing business. **[PUBLIC, Capital A FY2025 financial results]**

## The headline growth number needs context, not dismissal

FY2025 revenue reached **RM246 million, up 540% year-on-year**, with **EBITDA of RM118 million** and **NOP of RM95 million**. **[PUBLIC, Capital A FY2025 financial results]** A 540% growth figure this large is almost always a base-effect story rather than organic scaling, and that is exactly what happened here: AirAsia Next's FY2024 base was small precisely because it was still mostly a brand-licensing vehicle before BigPay's performance was folded in during FY2025. This does not make the current RM246 million revenue or RM95 million NOP any less real, but the 540% growth rate itself should not be read as evidence of an underlying business scaling at that pace; it is largely a reporting-structure change, similar in kind (though not in scale) to the entity-discontinuity caveats already established for the airline side of this project (Notebooks 3 and 5).

## 2026 growth rates have normalised, and the business remains genuinely profitable

With the FY2025 base effect behind it, AirAsia Next's 2026 growth rates look much more ordinary, and still positive. For **2Q2026**, revenue grew **8% year-on-year to RM78 million**, with **RM26 million EBITDA** and **RM22 million NOP**, a healthy 33% EBITDA margin. AirAsia Rewards specifically generated nearly **RM20 million in revenue (+20% YoY)**, with points redemptions up **34%**. On a first-half (1H2026) basis, revenue reached **RM166 million**, with **EBITDA of RM63 million** and **NOP of RM53 million**. **[PUBLIC, Capital A Financial Results Second Quarter 2026]** This is exactly what the base-effect explanation above predicts: once AirAsia Next was fully reorganised, its growth rate settled into the high single digits, in line with Capital A's other established businesses, rather than continuing anywhere near 540%.

## BigPay's own history, and the legal cloud sitting over its shares

Before its FY2025 fold-in, BigPay had its own separate track record: in FY2023 it reported **RM45 million revenue (+42% YoY)**, with its EBITDA loss narrowing 44% year-on-year, and it had previously drawn **up to USD100 million in financing from South Korea's SK Group** in August 2021, at the time positioned as "the first time any portfolio company within airasia Digital has secured financing of this size." **[PUBLIC, Notebook 2]**

That growth trajectory now sits alongside a genuine ownership risk detailed fully in Notebook 4: a Singapore High Court order (10 June 2026) to seize Move Digital's **99.56% stake in BigPay**, enforcing a **USD14.736 million** arbitration award to BigPay's minority co-founders. Whatever BigPay's operating performance shows inside AirAsia Next's FY2025 numbers, the entity's ownership itself is under active legal dispute, a risk that sits above the operating numbers rather than inside them.

## How this compares: a much larger, still-unprofitable regional fintech competitor

**Grab Holdings'** Financial Services segment, the closest comparable regional fintech business (GrabPay, lending, digital banking), reported **FY2025 revenue of USD347 million (+38% YoY)**, roughly **RM1.4-1.5 billion** at current exchange rates, about **6 times larger** than AirAsia Next's total FY2025 revenue. Notably, Grab's Financial Services segment was still **EBITDA-negative at USD-110 million** in FY2025, with the company targeting breakeven only in the second half of 2026. **[PUBLIC, Grab Holdings FY2025 results]**

This comparison is not fully apples-to-apples, since Grab's segment is pure fintech while AirAsia Next bundles brand licensing and loyalty alongside BigPay's fintech piece. But on the numbers available, AirAsia Next reports a profitable EBITDA at a fraction of the scale of a competitor that is still losing money on a much larger revenue base, a genuinely interesting divergence worth noting rather than assuming the bigger competitor is automatically doing better. **[DERIVED, this project's own comparison]**

## Explain It Simply

AirAsia Next started as a business that mostly just licensed the AirAsia brand name to other things. In 2025 it grew into a real operating company and absorbed BigPay, the fintech app, into its numbers, which is why its revenue looks like it grew 540% in one year, that's mostly an accounting reorganisation, not a business that actually got six times bigger overnight. BigPay itself has a real track record and real outside investment (from a Korean conglomerate), but it also has a genuine legal problem: a Singapore court has ordered Capital A's stake in it seized to settle an old dispute with two of its co-founders. Compared to Grab's much bigger digital-wallet business, which is still losing money at a much larger scale, AirAsia Next's smaller, profitable numbers are actually a reasonably good showing, legal cloud aside.

## Glossary

- **Brand licensing**: monetising a company's brand name and reputation by charging other businesses to use it, without directly operating those businesses.
- **Digital wallet**: a mobile app used to store money, make payments, and access other financial services (like BigPay).
- **Base effect**: when a very large percentage growth figure is driven mainly by a small starting number rather than genuinely fast underlying growth.

## Why Is This Happening?

**Why would Capital A fold a fintech app with an active legal dispute into a brand-licensing segment, rather than reporting it separately?** Grouping BigPay's fintech performance into AirAsia Next's broader loyalty-and-brand segment reflects a genuine strategic shift Capital A itself describes: building an "integrated fintech and loyalty platform," not simply a reporting convenience. Whether that consolidation makes commercial sense independent of the reporting choice is a separate question this project cannot fully answer from the segment figures alone, since BigPay's own standalone contribution within AirAsia Next's FY2025 numbers is not broken out separately in the company's disclosures.
