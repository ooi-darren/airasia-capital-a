**Business question**: Is AirAsia's crisis a company-specific problem, or is the whole low-cost carrier region under similar pressure? This notebook checks against three domestic and regional competitors, and is explicit about which comparisons the underlying data actually supports.

**Data note**: research before building this notebook found that Malaysia Airlines/Malaysia Aviation Group (MAG) is not separately listed (Khazanah-owned, no quarterly IR filings) and Batik Air Malaysia publishes no financial statements at all. ## VietJet: a full financial comparison

VietJet Aviation JSC publishes full audited annual reports. For FY2025 (year ended 31 December 2025): **net revenue VND82.09 trillion (+13.9% YoY)**, **net profit VND2.12 trillion (+51.3% YoY)**. Its Thai subsidiary, Thai Vietjet, grew its fleet 33% YoY (to 24 aircraft) with new 737-8 deliveries. **[PUBLIC, VietJet Aviation JSC 2025 Annual Report, audited consolidated financial statements]**

## Malaysia Aviation Group: annual figures only, and recovering

MAG (which operates Malaysia Airlines) is wholly owned by Khazanah Nasional and not separately listed, so no quarterly financial statements exist. Via its own press disclosures: **FY2025 net profit RM137 million** (revenue RM14.5 billion, +6%), its **fourth consecutive profitable year** post-pandemic, though it still holds RM1.77 billion of the RM3.6 billion in Khazanah funding extended since 2019, with no dividend yet paid back to its owner. **[DERIVED, compiled from press coverage: The Edge Malaysia, Free Malaysia Today, Malay Mail; MAG does not publish IR-style financial statements]**

The relevant read-through: Malaysia's flag carrier has quietly turned a corner into sustained profitability over the same period AirAsia, the country's dominant low-cost carrier, has been in acute distress, a genuinely interesting divergence in domestic aviation fortunes, not something either side of the AirAsia crisis narrative typically mentions.

## Batik Air Malaysia: capacity only, no financials exist

Batik Air Malaysia, a Lion Air Group joint venture, discloses no financial statements. What is publicly known: a fleet of **44 aircraft** (6 A330-300, 38 Boeing 737-8/800) serving **56 destinations**. **[PUBLIC, company/CAPA fleet data]** It is one of the two carriers the Malaysian government reportedly asked about absorbing AirAsia's domestic routes (Notebook 16), but there is no public data to assess whether it has the financial capacity to do so profitably; this is a genuine, disclosed information gap, not an oversight.

## Explain It Simply

To check whether AirAsia's troubles are just an industry-wide problem, this notebook looks at three competitors. VietJet, a similar low-cost carrier in Vietnam, actually grew its revenue and profit substantially over roughly the same period, real evidence this isn't purely a regional slump. Malaysia's own flag carrier, Malaysia Airlines, has quietly become profitable for four years running, the opposite direction from AirAsia. Batik Air, the other airline the government mentioned as a possible route-absorber, doesn't publish any financial information at all, so there's no way to check whether it's actually in good enough shape to take on more routes. Two out of three comparisons say AirAsia's problem is more company-specific than industry-wide; the third comparison simply can't be made with the data available.

## Glossary

- **MAG (Malaysia Aviation Group)**: the Khazanah Nasional-owned holding company operating Malaysia Airlines.
- **Khazanah Nasional**: Malaysia's sovereign wealth fund, the government's strategic investment arm.
- **JSC**: Joint Stock Company, VietJet's corporate form under Vietnamese law.

## Why Is This Happening?

**Why would VietJet and Malaysia Airlines both be doing well while AirAsia specifically struggles?** Part of the answer is structural, not just operational: both VietJet and Malaysia Airlines emerged from their own past restructurings (Vietnam's aviation sector consolidation, and MAS's 2014-2015 Khazanah-led restructuring into MAG) years before the 2026 fuel shock hit, giving them a longer runway of stability to build cash reserves and balance sheet strength. AirAsia Group, by contrast, only became a single consolidated entity in January 2026, just months before the fuel shock, meaning it entered this specific stress test with a brand-new, still-settling balance sheet rather than years of accumulated post-restructuring strength.
