# 03: From COVID Collapse to the Twin PN17 Sagas (2020-2026)

**Business question**: Is the September 2026 crisis AirAsia's first real brush with insolvency, or does the group have real, recent precedent for surviving something this serious, and for changing what it even is along the way? The answer changes how much weight to put on today's numbers versus the panic around them.

## The COVID-19 collapse: where this story actually starts

Before either PN17 classification, before any restructuring plan, there was an operational collapse severe enough to make both inevitable. When Malaysia's borders closed and global air travel stopped in March 2020, AirAsia's fleet was grounded almost in its entirety, essentially overnight, with no revenue to replace it.

The scale of the drop, comparing the same quarter one year apart: group revenue fell from **RM2.31 billion in Q1 FY2020** to **RM298.22 million in Q1 FY2021**, an **87% year-on-year fall**. The group posted a **net loss of RM767.42 million for that single quarter (Q1 FY2021)** alone. Across the full year, FY2020 revenue fell to **RM3.14 billion from RM11.86 billion**, and the group's net loss widened to **RM5.10 billion for the year**. **[PUBLIC]**

This was not a paper loss. By October 2020, AirAsia and AirAsia X together confirmed the retrenchment of roughly **10% of their combined 24,000-strong workforce, about 2,400 employees**, after an earlier June 2020 round had already cut 111 cabin crew, 172 pilots, and 50 engineers. **[PUBLIC]**

This is the collapse that both PN17 classifications below are a direct consequence of, not an unrelated prior chapter. ## Two entities, two separate distress classifications

PN17 (Practice Note 17) is Bursa Malaysia's classification for financially distressed listed companies, triggered when core financial ratios (commonly, shareholders' equity falling below 25% of share capital) breach set thresholds. Both AirAsia Group (the old parent, later renamed Capital A) and AirAsia X Berhad (the long-haul affiliate) triggered PN17 criteria around **30 July 2020**, amid COVID-driven travel collapse. Bursa's relief measures (April 2020 to June 2021) delayed formal classification for both. **[PUBLIC]**

From there, the two entities' paths diverged and were resolved separately, not as one combined process.

## AirAsia X's saga: the harsher one

AirAsia X was formally classified PN17 in **November 2021**, after its auditor issued a disclaimer of opinion on its accounts. By then, it had already proposed a **RM63.6 billion debt restructuring** (October 2020). Creditors approved the plan in November 2021 (100% of Class A and C creditors, 97.6% of Class B), and the High Court sanctioned it in December 2021. Under the approved scheme, creditors were paid only **0.5% of the RM33.65 billion** they were owed, a severe haircut that wiped the slate clean. AirAsia X formally exited PN17 status on **21 November 2023**. **[PUBLIC]**

## Capital A's saga: resolved through the merger itself

Capital A was formally classified PN17 in **January 2022**. Its own regularisation plan was not completed independently; it was completed **through** the January 2026 reverse acquisition itself: disposing its aviation businesses to the now debt-clean AirAsia X, distributing and listing AAX shares to entitled shareholders, and a High Court-approved capital reduction of approximately **RM5.5 billion**. Capital A formally exited PN17 on **20 May 2026**, closing a six-year restructuring chapter. **[PUBLIC]**

## Not just a restructuring: becoming Capital A

While the PN17 process ground on, the parent company also changed what it was. On **28 January 2022**, AirAsia Group Berhad renamed itself **Capital A Berhad**. This was not a cosmetic rebrand timed to a restructuring milestone; it was a public statement of strategic direction. Group CEO Tony Fernandes framed it directly: **"today's announcement reinforces we are not just an airline anymore."** **[PUBLIC]**

The rationale was a shift in the group's revenue mix, targeting roughly **50% of revenue from non-airline businesses by 2026**, up from the airline-dominated mix the group had run on for two decades. The businesses meant to carry that shift were already inside the group: **BigPay** (digital wallet and fintech, which secured up to **USD100 million in financing from South Korea's SK Group** in August 2021) and the **AirAsia super-app**, which had reached **10.6 million average monthly active users by Q2 2022** (up 236% year-on-year), aggregating flights, food delivery, and other services under one platform. The AirAsia brand itself was kept for the airlines; only the group holding company was renamed. **[PUBLIC]**

Read alongside the PN17 timeline below, the sequence is not coincidental: a company mid-restructuring, legally required to prove Bursa it had a credible path back to financial health, was simultaneously trying to convince investors that its future value lay outside the very airline business that had just needed to be rescued. Both things were true and were happening at once.

The sequence matters: AirAsia X had to become clean on its **own** balance sheet first (by November 2023) before it was in any position to serve as the vehicle that absorbed Capital A's aviation assets and completed Capital A's own regularisation. ## What this means for reading the September 2026 crisis

The group has direct, recent precedent for surviving something more severe than a bad quarter: a court-sanctioned restructuring that paid ordinary creditors half a cent on the ringgit, and it still exists, flies, and completed a six-year turnaround. That does not mean the current crisis is safe to dismiss; the working-capital deficit identified in Notebook 6 is real. It does mean this is not the group's first real test, and the institutional playbook (debt restructuring, court-sanctioned schemes, Bursa PN17 processes) has been used successfully before.

## Explain It Simply

COVID hit AirAsia harder than almost anything a normal business plan accounts for: revenue dropped 87% in a single year, the company lost hundreds of millions of ringgit in one quarter alone, and about 1 in 10 employees lost their jobs. That collapse is why both AirAsia's short-haul parent company and its long-haul sister airline, AirAsia X, were declared financially distressed by the stock exchange around the same time. But they were fixed separately. AirAsia X went through a brutal debt restructuring where the people it owed money to got paid less than one cent for every ringgit owed, and it took until late 2023 to come out the other side clean. In the middle of all this, the parent company also renamed itself Capital A and started telling investors its future was about more than flying planes, pointing to its digital wallet, cargo business, and super-app instead. Only after AirAsia X was clean could it be used, in January 2026, to absorb the parent company's airline business and finish cleaning up the parent's own books too. So this group has already been through a genuine near-death experience once, five years ago, survived it, and changed what it calls itself along the way. That's useful context for how seriously to take the current crisis.

## Glossary

- **PN17 (Practice Note 17)**: Bursa Malaysia's classification for listed companies in financial distress, typically triggered when shareholders' equity falls below a set threshold of share capital.
- **Regularisation plan**: the formal turnaround plan a PN17 company must submit and execute to be uplifted from distressed status.
- **Scheme creditors**: creditors bound by a court-sanctioned debt restructuring scheme, even those who did not individually vote for it, once the required majority approves.
- **Non-airline revenue**: revenue from businesses other than flying passengers, such as BigPay, Teleport, and the AirAsia super-app; the basis of Capital A's 2022 rebrand rationale.

## Why Is This Happening?

**Why would Capital A choose to resolve its own PN17 status through a merger rather than its own separate restructuring?** By 2025, AirAsia X was the cleaner, more attractive vehicle: a listed entity with no legacy debt overhang, precisely because of its own brutal 2021-2023 restructuring. Using it to absorb Capital A's aviation assets let Capital A satisfy Bursa's regularisation requirements while consolidating the group's flying operations into a single, better-capitalised entity, solving two problems (Capital A's PN17 status and the group's fragmented aviation structure) in one transaction rather than two separate ones.
