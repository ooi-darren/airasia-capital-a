## The business model, and a real mid-flight strategic change

Santan is Capital A's inflight catering and food brand, historically responsible for meal service on AirAsia flights, plus a growing retail and B2B food business. As of the most recent disclosure available to this project (1Q2026), Santan is executing a deliberate pivot: **stepping away from heavy inflight logistics**, moving to a **commission-based model with AirAsia starting May 2026**, and launching a **'Grab & Go' quick-service retail strategy** to grow its ground-based footprint, alongside a revamped beverage lineup. In the CEO's own words, this keeps the business "lean and agile, protecting operations from risk like food waste." **[PUBLIC, Capital A 1Q2026 financial results press release]**

## What the numbers show: fast early growth, now softening

In FY2023, Santan grew revenue to **RM132 million, up 276% year-on-year**, with EBITDA of **RM23 million**, a genuine turnaround from an EBITDA loss the prior year, though largely a post-COVID recovery story off a very depressed FY2022 base rather than a sustainable underlying growth rate. **[PUBLIC, Capital A FY2023 financial results]**

By 1Q2026, the picture had softened: quarterly revenue was **RM50.4 million, down 3.8% year-on-year**, attributed directly to "softer inflight activity following unplanned flight delays and low seasons following Ramadhan." Take-up rate (the share of passengers who buy something onboard) held flat at 29%, revenue per passenger fell 3.7% to RM3.66, and the average value per transaction (among passengers who did buy) fell 6% to RM12.21, two related but distinct measures both moving the same direction. **[PUBLIC, Capital A 1Q2026 financial results press release]** The company's own disclosure frames this softness explicitly as a reason for the strategic pivot described above, moving revenue risk off Santan's own books via the commission-based model, rather than continuing to bear inflight demand volatility directly. This is Capital A's second business, alongside AirAsia MOVE (Notebook 10), where the airline's own operating conditions (fewer flights, disrupted schedules) directly compress a nominally separate business's results.

## 2Q2026: revenue stabilised, but volume kept falling

For **2Q2026**, Santan reported revenue of **RM45 million, flat year-on-year on a normalised basis**, with **EBITDA of RM2 million** and **NOP of RM1 million**, Capital A's thinnest margin of the five businesses that quarter (Notebook 13). Passenger volume served fell a further **14% year-on-year**, meaning flat revenue was achieved despite fewer passengers to sell to, consistent with the pricing and mix actions described in the strategic pivot above rather than a genuine demand recovery. On a first-half (1H2026) basis, revenue softened **2% year-on-year to RM95 million**, with EBITDA and NOP narrowing to **RM4 million** and **RM2 million** respectively. **[PUBLIC, Capital A Financial Results Second Quarter 2026]** This project's most recent data does not confirm whether the commission-based model or the "Grab & Go" retail launch, both planned to begin around May 2026, had started contributing to these 2Q26 figures.

## Competitors: no usable comparison exists

Airline catering is typically run as an internal cost center by other carriers rather than a separately reported, market-facing business, so there is no comparable publicly-disclosed competitor to check Santan against, the same kind of genuine data gap already disclosed for Batik Air (Notebook 7) and AirAsia MOVE (Notebook 10). Santan's own pivot toward third-party retail ("Grab & Go") is itself an attempt to enter a market (quick-service food retail) with real, disclosed competitors, but it had not yet launched as of this project's most recent data. **[Disclosed limitation]**

## Explain It Simply

Santan makes the food AirAsia serves on its planes, plus a smaller retail and B2B business. It grew explosively right after COVID as travel came back, but by early 2026 its inflight revenue was softening because AirAsia itself was having a rough patch with flight delays and quieter travel seasons, and by mid-2026 revenue had gone flat while the actual number of passengers it served kept falling. Rather than keep absorbing that risk directly, Santan's management is deliberately stepping back from running the heavy logistics of inflight catering and shifting to a simpler arrangement where AirAsia pays it a commission instead, while Santan focuses its energy on a new fast-food retail concept called 'Grab & Go' to sell food directly to customers on the ground, not just on planes. It's too early to say whether that pivot is working, since this project's most recent numbers don't yet show whether it had even started.

## Glossary

- **Take-Up-Rate (TUR)**: the share of passengers who actually purchase inflight food or beverages, out of all passengers offered the chance to.
- **Commission-based model**: an arrangement where one party pays another a percentage of revenue or a fee per transaction, rather than the second party bearing the full cost and risk of running the underlying operation itself.

## Why Is This Happening?

**Why would Santan give up direct control of inflight catering rather than just cutting costs within the existing model?** Inflight catering carries structural risks (food waste, logistics complexity, exposure to flight schedule disruptions) that scale with how much of the airline's actual flying activity is disrupted, exactly what happened in 1Q2026. Moving to a commission-based arrangement transfers that operational risk to a simpler, more predictable revenue stream tied to whatever AirAsia actually flies, while freeing Santan's management attention to build a ground-based retail business (Grab & Go) whose success does not depend on the airline's own capacity decisions, directly addressing the same coupling risk already identified for AirAsia MOVE (Notebook 10).
