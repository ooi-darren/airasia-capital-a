**Business question**: Teleport is Capital A's single largest revenue contributor and the business most directly caught in a legal dispute over its own shares (Notebook 4). Is it actually a good business on its own merits, and how does it stack up against dedicated logistics competitors?

## The business model: asset-light, capacity-blended

Teleport moves cargo and e-commerce parcels by combining three capacity sources: **AirAsia Group's own belly space** (the cargo hold beneath passenger flights), **third-party partner airline capacity** (over 30 airline partnerships), and **its own freighter aircraft** (three Airbus A321 Freighters as of FY2023, since expanded). This asset-light structure means Teleport's growth is not capped by how many freighters it owns outright. **[PUBLIC, Capital A financial results press releases]**

## Three years of real trend: fast growth, then a real turnaround

Revenue grew from **RM731 million (FY2023)** to **RM1.1 billion (FY2024, +49% YoY)** to **RM1.2 billion (FY2025, +11% YoY)**, a genuine deceleration in growth rate as the base got larger, not a red flag on its own. The more interesting story is the bottom line: FY2024 closed with a **RM21 million net loss** despite EBITDA growing to RM90 million that year (interest and depreciation on the growing freighter fleet ate the difference), before Teleport achieved a **RM18.6 million Net Operating Profit in FY2025**, a genuine RM40 million swing, driven partly by refinancing its Deutsche Bank loan in 3Q2025 to lower finance costs. **[PUBLIC, Capital A FY2023-FY2025 financial results press releases]**

Volume tells the same growth story from the operating side: FY2025 total volume reached **347,885 tonnes (+18% YoY)** and **167 million parcels moved (+99% YoY)**, with the parcel count roughly doubling even as tonnage grew more modestly, a shift toward higher-frequency, smaller e-commerce shipments over bulk cargo. **[PUBLIC]**

Within Capital A's own portfolio, Teleport is the anchor: it and ADE together generated over 70% of 1H26 group revenue, with Teleport alone about 40% of Capital A's Q1 2026 revenue (Notebook 4).

## The turnaround is continuing into 2026, not a one-off

The FY2025 turnaround was not a one-time event. For **2Q2026**, Teleport reported **revenue of RM311 million (+22% YoY)**, **EBITDA of RM26.8 million (+9% YoY)**, its **strongest 2Q EBITDA to date**, and **PAT of RM4.4 million**, its **fourth consecutive profitable quarter**. On a first-half (1H2026) basis, revenue reached **RM620 million (+21% YoY)**, and PAT turned to **+RM12.2 million**, against an actual **RM1.5 million loss in 1H2025**, a genuine swing back to profitability rather than a single strong quarter. **[PUBLIC, Capital A Financial Results Second Quarter 2026]**

## How this compares: a real competitor, but a very different scale

**J&T Express**, a publicly-reporting regional logistics company, is a legitimate comparison since both move parcels across Southeast Asia, though its core model (ground-based parcel delivery network) differs structurally from Teleport's air-cargo, belly-space-blended model. For FY2025, J&T Express's Southeast Asia segment alone reported **US$4.5 billion in revenue (+39.8% YoY)** on **7.66 billion parcels (+67.8% YoY)**. **[PUBLIC, J&T Express FY2025 results]**

At roughly RM1.2 billion (about USD260-280 million depending on the exchange rate used), Teleport's revenue is on the order of **15-17 times smaller** than J&T's Southeast Asia segment alone, and its 167 million annual parcels are a small fraction of J&T's 7.66 billion in the same region. This is not a fair apples-to-apples read on execution quality, since the business models differ (air cargo and belly-space brokering versus a dedicated ground parcel network), but it is an honest scale check: Teleport is a genuinely fast-growing niche player within a market where a dedicated regional logistics operator is more than an order of magnitude larger. **[DERIVED, this project's own comparison]**

## Explain It Simply

Teleport is Capital A's cargo and delivery business, and it doesn't own a huge fleet of its own, it mostly borrows space in the belly of AirAsia's passenger planes, buys capacity from other airlines, and only owns a few dedicated freighter planes. Its revenue nearly doubled over two years, but in 2024 it actually lost money on the bottom line even while its core operating profit grew, mostly due to interest costs on debt. In 2025 it fixed that, refinanced its debt, and turned an actual profit for the first time. Compared to J&T Express, a real logistics company operating across the same region, Teleport is still 15-20 times smaller, so while it's Capital A's star performer internally, it's a small player in the wider regional logistics market.

## Glossary

- **Belly space**: the cargo hold underneath a passenger aircraft's main deck, used to carry freight alongside passengers.
- **Freighter aircraft**: an aircraft configured to carry cargo only, with no passenger seating.
- **NOP (Net Operating Profit)**: a Capital A-specific measure of core operating profit, excluding certain one-off items.

## Why Is This Happening?

**Why did EBITDA grow strongly in FY2024 while the bottom line still showed a loss?** EBITDA excludes interest and depreciation, the exact costs that scale up when a business finances growth (new freighters, expanded operations) with debt. Teleport's FY2024 pattern, rising EBITDA alongside a net loss, is a common signature of a business scaling capital-intensive infrastructure faster than it can pay down the associated financing costs. The FY2025 turnaround came specifically from addressing the financing side (refinancing the Deutsche Bank loan) rather than from a fundamental change in the underlying operating business, which was already healthy in FY2024.
