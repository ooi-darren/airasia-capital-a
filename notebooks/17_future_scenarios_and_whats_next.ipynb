## The two levers that matter most

This project's own findings point to two variables that plausibly determine which direction this goes, more than any single announcement:

1. **Whether the refinancing succeeds.** AirAsia is seeking up to **USD1 billion** in international debt markets plus **RM700 million** in local credit facilities, explicitly for debt refinancing and balance-sheet consolidation, alongside a separate attempt to amend terms on an existing **USD200 million** private credit loan. **[PUBLIC]** Notebook 14's RM14.5 billion working-capital deficit is the gap this financing is meant to address, though as the next section shows, the two are not the same size.
2. **Whether the fuel price shock persists or eases.** Notebook 5 showed the operating loss is fuel-driven, not a broad cost failure. A retreat in fuel prices would substantially improve the picture without any restructuring action at all; continued high prices would keep pressure on regardless of financing outcomes.

## Does the financing being sought actually close the gap it's meant to close?

This is a question this project's own notebooks never put to each other directly until now: Notebook 14 identified a **RM14.5 billion** working capital deficit; this notebook's lever 1 is the **USD1 billion plus RM700 million** AirAsia is seeking. Are those actually comparable in size?

```python
# AirAsia's own disclosed 2Q26 average exchange rate, for a consistent conversion basis
usd_myr = 3.99

deficit_rm_bn = 14.544
financing_sought_usd_bn = 1.0
financing_sought_rm_bn = 0.700
financing_sought_usd_equiv_bn = financing_sought_usd_bn + (financing_sought_rm_bn / usd_myr)

deficit_usd_equiv_bn = deficit_rm_bn / usd_myr
coverage_pct = financing_sought_usd_equiv_bn / deficit_usd_equiv_bn

print(f"Working capital deficit (Notebook 14): RM{deficit_rm_bn:.2f}bn = USD{deficit_usd_equiv_bn:.2f}bn equivalent")
print(f"Financing being sought (Notebook 17's lever 1): USD{financing_sought_usd_bn:.1f}bn + RM{financing_sought_rm_bn:.2f}bn = USD{financing_sought_usd_equiv_bn:.2f}bn equivalent")
print(f"That covers roughly {coverage_pct:.0%} of the diagnosed deficit, at face value.")
```

On the numbers alone, the financing being sought covers only about **a third** of the working capital deficit this project's own Notebook 14 diagnosed. Three caveats keep this from being as alarming as it first looks. First, "seeking" is not "secured": these are amounts AirAsia says it is pursuing, not financing it has already closed, and the gap could look different again once (if) deals are signed. Second, not all of a working capital deficit needs to be closed with fresh external financing: Notebook 14 itself established that airlines structurally run negative working capital, since ticket cash comes in before the flight is flown, so part of the RM14.5 billion self-resolves through ongoing sales rather than requiring a matching cash injection. Third, Notebook 14's cash flow analysis above shows the Group is already actively rolling roughly RM2.8 billion of financing every six months; the USD1 billion plus RM700 million is additive to that ongoing activity, not the first or only lever being pulled.

None of that makes the mismatch disappear. Even generously discounting the deficit for its structural, self-resolving portion, a financing package sized at roughly a third of the headline gap is either meant to bridge only part of it (in which case something else, cost cuts, asset sales, or the capital reduction exercise already underway, needs to close the rest) or the deficit itself is not the number the financing is actually being sized against. Neither this project nor any source it draws on says which. **[DERIVED, this project's own calculation from figures established in Notebooks 6 and 11]**

## What this project's own evidence actually weighs toward, including where it disagrees with itself

A first pass at this section leaned entirely toward the optimistic end of the range. On review, that reading cherry-picked the favorable half of two facts this project's own earlier notebooks left genuinely unresolved, so it is corrected here rather than left standing.

**The recency of the merger cuts both ways, and this project used only one side of it.** Notebook 2 cites AirAsia X's successful 2020-2023 restructuring as reason to think this group can survive a serious crisis. But Notebook 7's own explanation for why AirAsia Group is underperforming VietJet and MAG right now is that it "only became a single consolidated entity in January 2026... a brand-new, still-settling balance sheet rather than years of accumulated post-restructuring strength." Both claims come from this project, and they point in opposite directions from the same underlying fact: the entity is new. A fair synthesis has to hold both at once, past institutional resilience exists, but it belongs to AirAsia X specifically, not yet to the untested combined AirAsia Group, rather than quietly using only the flattering half.

**The government's lighter-touch response is genuinely ambiguous, and this project should not resolve that ambiguity by omission.** Notebook 16 explicitly presented two readings, a milder assessment of severity, or simply the absence of an ownership lever like the one Khazanah held over MAS in 2014, and stated it does not know which explains the lighter response. Treating the lighter-touch response as evidence of lower severity, without carrying forward that same hedge, overstates what Notebook 16 actually supports.

**What can honestly be said**: the RM14.5 billion working capital deficit (Notebook 14) is real and does not depend on any of the above being resolved either way. The fuel-driven, not-mismanagement read of the 2Q26 loss (Notebook 5) is comparatively solid, since it rests on the company's own disclosed CASK ex-fuel improvement, not on an interpretation of ambiguous signals. The AirAsia X precedent (Notebook 2) and the government's posture (Notebook 16) are both real data points, but neither is a strong reason for optimism on its own once their own stated ambiguity is taken seriously; they are pieces of context, not evidence that resolves the outcome.

## Explain It Simply

Nobody, including this project, knows for certain what happens next to AirAsia. What can be said is that two things mostly decide it: whether AirAsia can raise the money it's trying to raise, and whether jet fuel prices come back down. If both go AirAsia's way, this crisis probably fades quietly. If neither does, the more severe scenarios discussed in the news, like other airlines taking over its routes, become more realistic.

An earlier version of this notebook argued the overall balance of evidence leaned toward the less alarming outcome, using AirAsia X's past recovery and the government's calm-seeming response as support. On a closer look, that argument was too generous to itself: this same project's own Notebook 7 explains AirAsia's current underperformance by pointing to exactly the thing Notebook 2 used as a reason for confidence, that the merged company is brand new. And Notebook 16 said plainly it couldn't tell whether the government's light touch means low risk or simply that it has no ownership stake to leverage. The one finding that doesn't depend on picking a favorable interpretation is that the cash crunch is real and the loss is fuel-driven rather than self-inflicted. Beyond that, this project genuinely does not have strong enough evidence to lean toward either the calm or the alarming outcome, and says so plainly rather than picking a side to sound more conclusive.

## Glossary

- **Scenario analysis**: laying out a structured range of plausible futures based on identified variables, explicitly not the same as a forecast or probability estimate.

## Why Is This Happening?

**Why build scenarios instead of just picking the most likely one?** A single predicted outcome would misrepresent the actual state of the evidence: this is a live situation where the two decisive variables, refinancing success and fuel prices, are both genuinely uncertain and outside any single party's full control. Presenting a range tied explicitly to this project's own findings, rather than a single confident guess, is the more honest and more useful output for a reader trying to assess exposure or make a decision, exactly the standard this portfolio has held to since Notebook 1.
