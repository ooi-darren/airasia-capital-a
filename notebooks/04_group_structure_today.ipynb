## Capital A Berhad: the original mother, now the non-aviation holding company

Capital A is the continuous legal entity founded in 1993 and bought for RM1 in 2001 (Notebook 1), renamed from AirAsia Group Berhad in January 2022 (Notebook 2). After disposing of its aviation businesses (completed December 2025-January 2026, Notebook 3), it retained five core businesses, reported as its own segments: **ADE** (aircraft maintenance, repair and overhaul), **Teleport** (air and ground logistics/cargo), **AirAsia MOVE** (the travel booking super-app, formerly airasia Superapp), **AirAsia Next** (brand licensing and loyalty, which now also houses **BigPay**, the fintech app), and **Santan** (inflight catering and food retail). For FY2025, Capital A Companies generated **RM3.39 billion in revenue**, **RM443 million EBITDA**, and **RM171 million NOP**, with a one-off gain on the aviation disposal restoring the Group to **positive shareholder equity of RM937 million**. **[PUBLIC, Capital A FY2025 financial results]**

## AirAsia Group Berhad: legally the former sister company, now the airline holding company

The company now called AirAsia Group Berhad is, legally, the entity that used to be called AirAsia X Berhad, the long-haul affiliate founded in 2006/2007 (Notebook 1). It legally acquired Capital A's short/medium-haul airlines on 16 January 2026 (a reverse acquisition, Notebook 3) and renamed itself AirAsia Group Berhad on 2 July 2026, reusing the original parent's old name. It now consolidates six Air Operator Certificates: **Malaysia AirAsia (MAA)**, **Malaysia AirAsia X (MAAX)**, **Thai AirAsia (TAA)**, **Indonesia AirAsia (IAA)**, **Philippines AirAsia (PAA)**, and **Cambodia AirAsia (CAA)**, plus **Thai AirAsia X (TAAX)**, held as an equity-accounted associate rather than a consolidated subsidiary. For 2Q26, the consolidated group reported **RM5.09 billion in quarterly revenue**. **[PUBLIC, AirAsia Group 2Q26 Quarterly Announcement]**

This entity, and its crisis, is the subject of Notebooks 5-7 and 14 (the airline's business model, fleet, competitive position, and the September 2026 liquidity crisis).

## The link between them: a 19.5% stake, not a parent-subsidiary relationship

Capital A is confirmed as AirAsia Group's **single largest shareholder, holding a 19.5% stake**, received as consideration for handing over its aviation businesses rather than cash. **[PUBLIC, Capital A Major Shareholders disclosure]** This means Capital A retains real economic exposure to AirAsia Group's fortunes (quantified directly in Notebook 15) without controlling it and without its results being consolidated into Capital A's own financial statements. Anyone reading a headline number for "AirAsia" without checking which of the two companies it refers to risks mixing up two different balance sheets, two different sets of subsidiaries, and two different sets of problems.

## A live complication in Capital A's own ownership structure

On **10 June 2026**, the Singapore High Court ordered the seizure of Move Digital Sdn Bhd's (a Capital A subsidiary) **99.56% stake in BigPay** and **11.45% stake in Teleport**, enforcing a **27 December 2024** partial arbitration award. The dispute traces back to 2021, when BigPay's minority co-founders, Christopher Davison and Navin Rajagopalan, sought a buyout of their stakes under a 2017 shareholder agreement, claiming between **US$140-183 million**; the arbitration tribunal awarded them only **US$14.736 million (approximately RM60.09 million)**. Capital A/Move Digital filed objections to the seizure, which remain pending as of this project's research. **[PUBLIC, AirAsia Group 2Q26 Quarterly Announcement, Note 33]**

This sits squarely inside Capital A's own ownership structure, not AirAsia Group's, and predates the September 2026 aviation crisis by three months. It is a genuine, independent risk factor sitting over exactly two of the five businesses (BigPay and Teleport) covered in this project's business-unit notebooks.

## Explain It Simply

There isn't one "AirAsia" anymore, there are two separate public companies that both grew out of the same 1993 airline. Capital A is the original company, renamed, and it now owns the logistics arm, the aircraft maintenance business, the travel app, the loyalty and fintech brand, and the catering company, everything except the actual planes. AirAsia Group is, confusingly, the company that used to be called AirAsia X, which bought the actual airlines from Capital A in 2026 and then took over the old "AirAsia Group" name for itself. Capital A kept a 19.5% ownership slice of AirAsia Group as part of that deal, so the two are financially linked, but neither owns the other outright. On top of that, a Singapore court has ordered two of Capital A's own businesses, BigPay and part of Teleport, seized to settle an old legal dispute, a separate complication sitting inside Capital A's own house.

## Glossary

- **AOC (Air Operator Certificate)**: the regulatory licence required to operate an airline; AirAsia Group consolidates six of these.
- **Associate**: a company in which an investor holds significant influence (commonly 20-50% ownership) but not control, accounted for differently from a consolidated subsidiary; Thai AirAsia X (TAAX) is held this way.
- **Cross-shareholding**: when one company owns a minority equity stake in another, creating financial exposure without control or consolidation.
- **Arbitration award**: a binding decision from a private arbitration tribunal, enforceable through the courts like a judgment.
